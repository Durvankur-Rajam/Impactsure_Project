```python
#GPU_Time

from transformers import pipeline
import time

print("Loading model on GPU...")
generator = pipeline("text-generation", model="gpt2", device=0)  # device=0 = GPU

prompt = "Artificial intelligence will change the world by"

start = time.time()
output = generator(prompt, max_length=100, num_return_sequences=1)
end = time.time()

print("\n--- Generated Text ---")
print(output[0]['generated_text'])
print(f"\n⏱ Time taken (GPU): {end - start:.2f} seconds")
```

```python
#CPU_Time

from transformers import pipeline
import time

print("Loading model on CPU...")
generator_cpu = pipeline("text-generation", model="gpt2", device=-1)  # CPU

prompt = "Artificial intelligence will change the world by"

start = time.time()
output_cpu = generator_cpu(prompt, max_length=100, num_return_sequences=1)
end = time.time()

print("\n--- Generated Text (CPU) ---")
print(output_cpu[0]['generated_text'])
print(f"\n⏱ Time taken (CPU): {end - start:.2f} seconds")
```

```python
from transformers import pipeline
import time

prompt = "Artificial intelligence will change the world by"

print("Running on GPU...")
generator_gpu = pipeline("text-generation", model="gpt2", device=0)

start = time.time()
output_gpu = generator_gpu(prompt, max_length=100, num_return_sequences=1)
gpu_time = time.time() - start

print("\n--- Generated Text (GPU) ---")
print(output_gpu[0]['generated_text'])
print(f"⏱ GPU Time: {gpu_time:.2f} seconds")

print("\nRunning on CPU...")
generator_cpu = pipeline("text-generation", model="gpt2", device=-1)

start = time.time()
output_cpu = generator_cpu(prompt, max_length=100, num_return_sequences=1)
cpu_time = time.time() - start

print("\n--- Generated Text (CPU) ---")
print(output_cpu[0]['generated_text'])
print(f"⏱ CPU Time: {cpu_time:.2f} seconds")

print("\n" + "=" * 40)
print("        GPU vs CPU COMPARISON")
print("=" * 40)
print(f"GPU Time : {gpu_time:.2f} seconds")
print(f"CPU Time : {cpu_time:.2f} seconds")
print(f"Speedup  : {cpu_time / gpu_time:.1f}x faster on GPU")
```
