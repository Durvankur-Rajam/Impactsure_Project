TruthfulQA Summary

Overview

TruthfulQA (Lin et al., 2021) is a benchmark of 817 adversarially designed questions across 38 categories (health, law, finance, conspiracies, etc.) that test whether LLMs generate truthful answers. Questions target areas where humans commonly hold false beliefs.

Key Findings

Larger models were LESS truthful — they better absorb human misconceptions from training data (Inverse Scaling)
Best model answered only ~58% of questions truthfully vs ~94% for humans
Models most commonly failed on health myths, legal misconceptions, and popular fiction
RLHF-tuned models (InstructGPT) were more truthful than base GPT-3
Automated evaluation using a fine-tuned GPT-judge achieved ~90% agreement with human raters

Relevance to This Project

TruthfulQA directly provides the 817-question benchmark (already loaded in Task 8) to evaluate LLMs for hallucinations. It defines hallucination as asserting a false statement learned from training data — the foundation for this project.

SelfCheckGPT Summary


Overview

SelfCheckGPT (Manakul et al., 2023) introduces a zero-resource method to detect hallucinations in LLM outputs without needing external databases or ground truth. The core idea: if a model truly knows a fact, it will consistently reproduce it across multiple samples. Inconsistency = hallucination.


How It Works

Generate multiple responses to the same prompt from the model (e.g., 5–20 samples)
Check each sentence in the main response for consistency across all sampled responses
Sentences that appear consistently across samples are likely factual
Sentences that are contradicted or absent in other samples are flagged as hallucinations
No external knowledge base needed — the model checks itself (hence 'Self-Check')

Consistency Checking Methods

BERTScore: Semantic similarity between sentences across samples
NLI (Natural Language Inference): Checks if samples entail or contradict the sentence
n-gram overlap: Simple word overlap as a lightweight alternative

Key Findings

SelfCheckGPT strongly correlates with human hallucination judgements on WikiBio dataset
NLI-based checking performed best overall
Works on black-box models no access to model weights or logits required
