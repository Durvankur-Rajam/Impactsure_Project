```python
!pip install transformers -q

from transformers import GPT2LMHeadModel, GPT2Tokenizer
import torch
import time

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")
model = model.to("cuda")

print("Model loaded on:", next(model.parameters()).device)
```

```python
prompt = "The future of artificial intelligence is"
input_ids = tokenizer.encode(prompt, return_tensors="pt").to("cuda")

temperatures = [0.2, 0.7, 1.5]
results = {}

for temp in temperatures:
    start = time.time()

    output = model.generate(
        input_ids,
        max_length=100,
        temperature=temp,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )

    elapsed = time.time() - start
    decoded = tokenizer.decode(output[0], skip_special_tokens=True)
    results[temp] = {"text": decoded, "time": elapsed}

    print(f"\n{'='*50}")
    print(f"🌡 Temperature: {temp}")
    print(f"⏱ Time: {elapsed:.2f}s")
    print(f"📝 Output:\n{decoded}")
```

```python
print("\n" + "="*60)
print(f"{'TEMPERATURE COMPARISON TABLE':^60}")
print("="*60)
print(f"{'Temp':<8} {'Time (s)':<12} {'First 60 chars of output'}")
print("-"*60)

for temp, data in results.items():
    preview = data['text'][:60].replace('\n', ' ')
    print(f"{temp:<8} {data['time']:<12.2f} {preview}...")

print("="*60)
print("\nObservations:")
print("T=0.2 → Focused, repetitive, predictable")
print("T=0.7 → Balanced, coherent, natural")
print("T=1.5 → Creative, random, sometimes incoherent")
```
