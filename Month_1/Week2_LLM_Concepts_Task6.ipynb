## 1. What is a Large Language Model (LLM)?

A Large Language Model (LLM) is a deep learning model trained on massive amounts of text data to understand and generate human language. LLMs learn statistical patterns across billions of words, allowing them to predict, complete, translate, summarise, and reason over text.

**Key characteristics:**
- Trained on large-scale text corpora (books, websites, code, articles)
- Use the Transformer architecture as their backbone
- Scale is critical — more parameters generally means better performance
- Examples: GPT-2, GPT-4, LLaMA 3, Mistral, Claude, Gemini

## 2. Attention Mechanism

Attention is the core innovation of the Transformer architecture (Vaswani et al., 2017). It allows the model to focus on relevant parts of the input when generating each word.

**How it works:**
For each token generated, attention computes a weighted score over all other tokens. More relevant tokens get higher weights.

**Example:**
Sentence: "The bank by the river was flooded."
When predicting "bank", attention assigns high weight to "river" and "flooded" — correctly identifying it as a riverbank, not a financial institution.

**Types:**
- Self-attention: Each token attends to every other token in the same sequence
- Multi-head attention: Multiple attention operations run in parallel, each learning different relationships (syntax, semantics, coreference)


## 3. Tokens

A token is the basic unit of text that an LLM processes. Tokens are not always full words — they can be words, subwords, punctuation, or characters.

| Text           | Tokens                          | Count |
|----------------|---------------------------------|-------|
| Hello, world!  | ['Hello', ',', ' world', '!']  | 4     |
| Artificial     | ['Art', 'ific', 'ial']         | 3     |
| unhappiness    | ['un', 'happiness']            | 2     |

**Why tokens matter:**
- LLMs have a context window limit (GPT-2: 1024 tokens, GPT-4: 128k tokens)
- Longer inputs cost more compute and may exceed the limit
- API pricing is measured per token

## 4. Temperature

Temperature controls the randomness of an LLM's output during generation.

Formula: P(token) = softmax(logits / T)

| Temperature | Effect                                      | Best Used For                     |
|-------------|---------------------------------------------|-----------------------------------|
| T = 0.0     | Deterministic — always picks highest prob   | Factual Q&A, classification       |
| T = 0.1–0.5 | Low randomness — focused output             | Summarisation, translation        |
| T = 0.7–1.0 | Balanced — coherent with some creativity    | General chat, writing             |
| T = 1.2–2.0 | High randomness — creative but incoherent   | Brainstorming, poetry             |

**Temperature & Hallucinations:**
Higher temperatures increase the chance of selecting low-probability tokens, which can lead to hallucinations — factually incorrect or fabricated outputs.
